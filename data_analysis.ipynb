```python
import json
from collections import Counter
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

json_path = Path("celebvhq_info.json")
if not json_path.exists():
    json_path = Path("/home/sh0408sh/FaceGeneration/CelebV-HQ/celebvhq_info.json")

with open(json_path, "r") as f:
    data = json.load(f)

ytb_ids = [v["ytb_id"] for v in data["clips"].values()]
counter = Counter(ytb_ids)
clip_counts = np.array(list(counter.values()))

summary_df = pd.DataFrame(
    {
        "metric": [
            "num_youtube_videos",
            "num_clips",
            "mean_clips_per_video",
            "median_clips_per_video",
            "max_clips_per_video",
        ],
        "value": [
            len(clip_counts),
            int(clip_counts.sum()),
            round(float(clip_counts.mean()), 2),
            int(np.median(clip_counts)),
            int(clip_counts.max()),
        ],
    }
)

dist_df = (
    pd.Series(clip_counts)
    .value_counts()
    .sort_index()
    .rename_axis("clips_per_youtube_video")
    .reset_index(name="num_youtube_videos")
)
dist_df["ratio"] = dist_df["num_youtube_videos"] / dist_df["num_youtube_videos"].sum()
dist_df["ratio_percent"] = (dist_df["ratio"] * 100).round(2)

print(f"metadata path: {json_path}")
display(summary_df)
display(dist_df)

bins = np.arange(0.5, clip_counts.max() + 1.5, 1)

plt.figure(figsize=(10, 5))
plt.hist(clip_counts, bins=bins, edgecolor="black", alpha=0.8)
plt.title("Number of Clips per YouTube Video")
plt.xlabel("clips per YouTube video")
plt.ylabel("number of YouTube videos")
plt.grid(axis="y", alpha=0.3)

# Clip the x-axis for readability when a few videos have many clips.
if clip_counts.max() > 30:
    plt.xlim(0.5, 30.5)
    plt.xticks(range(1, 31))
    plt.text(
        0.98,
        0.95,
        f"x-axis clipped at 30; max={clip_counts.max()}",
        transform=plt.gca().transAxes,
        ha="right",
        va="top",
    )
else:
    plt.xticks(range(1, clip_counts.max() + 1))

plt.tight_layout()
plt.show()

# Videos with the largest number of clips.
top_videos_df = pd.DataFrame(counter.most_common(20), columns=["ytb_id", "num_clips"])
display(top_videos_df)
```

```python
import pandas as pd

rows = []

for clip_id, val in data["clips"].items():
    rows.append({
        "clip_id": clip_id,
        "ytb_id": val["ytb_id"],
        "start_sec": val["duration"]["start_sec"],
        "end_sec": val["duration"]["end_sec"],
        "duration": val["duration"]["end_sec"] - val["duration"]["start_sec"],
        "bbox_top": val["bbox"]["top"],
        "bbox_bottom": val["bbox"]["bottom"],
        "bbox_left": val["bbox"]["left"],
        "bbox_right": val["bbox"]["right"],
    })

df = pd.DataFrame(rows)
df.head()
```
